````python
import numpy as np
import matplotlib.pyplot as plt

# === EXPONENTIAL DAMPING PROXY (Hubble + Logosfield) ===
t = np.linspace(0, 700, 1000)  # Myr
H = 0.01
gamma = 0.0005
r0 = 3.53

dist_no = r0 * np.exp(H * t)
dist_field = r0 * np.exp((H - gamma) * t)

final_no = dist_no[-1]
final_field = dist_field[-1]
damping = (final_no - final_field) / final_no * 100

print(f"ΛCDM final mean dist: {final_no:.4f}")
print(f"Logosfield final mean dist: {final_field:.4f}")
print(f"Damping: {damping:.1f}%")

plt.figure(figsize=(10,6))
plt.plot(t, dist_no, 'r--', label='ΛCDM (No Field)', linewidth=2)
plt.plot(t, dist_field, 'b-', label='Logosfield (γ=0.005)', linewidth=2.5)
plt.axhline(final_no, color='r', ls=':', alpha=0.7)
plt.axhline(final_field, color='b', ls=':', alpha=0.7)
plt.title('Logosfield Damping: 29% Tighter Clusters at z>10', fontsize=14)
plt.xlabel('Time (Myr)')
plt.ylabel('Mean Distance (arbitrary units)')
plt.legend()
plt.grid(alpha=0.3)
plt.text(350, final_field*1.1, f'{damping:.1f}% Damping', 
         bbox=dict(boxstyle="round", facecolor="lightblue"))
plt.tight_layout()
plt.savefig('logosfield_damping_plot.png', dpi=300)
plt.show()
````
