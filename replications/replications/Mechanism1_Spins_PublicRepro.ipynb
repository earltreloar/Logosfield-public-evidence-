# Mechanism 1 — Galaxy Spin Alignment (Public Reproduction)

Goal:
- Reproduce the pooled galaxy spin alignment fraction `f_align` ~0.62–0.64.
- Show that this is 
  - (a) not consistent with isotropic nulls,
  - (b) persists across surveys (SDSS, JWST high-z bins),
  - (c) lines up with the *same* Logosfield scalar parameters (α=1, β=2π, γ=0.005) used in other mechanisms.

This notebook is the minimal rerun. It uses the pre-extracted spin catalog
`mechanism1_all_surveys_alignment_vs_z_3.1.1.csv` from the public evidence pack.

Headline claims to verify:
* pooled alignment fraction f ≈ 0.625 over N ≈ 1.3M spins, p < 1e-6.
* SDSS-only ~0.623.
* JWST high-z bins ~0.613–0.693.

These are the values cited in the v2.0 README and rebuttal.

```python
import pandas as pd
import numpy as np

# Path assumption: the CSV from the release bundle lives alongside this repo
# If running in Colab, upload that CSV and set csv_path accordingly.
csv_path = "mechanism1_all_surveys_alignment_vs_z_3.1.1.csv"

df = pd.read_csv(csv_path)

# Expected columns (typical in your bundle):
# - survey  (e.g. 'SDSS', 'JWST', etc.)
# - z_bin   (redshift bin label)
# - N       (count of galaxies in that bin)
# - f_align (fraction aligned with reference axis / Logosfield axis)
# - f_null_mean, f_null_std  (null model from shuffle/rotation)

print("Columns:", list(df.columns))
df.head()
```

```python
# Compute pooled alignment fraction and a simple z-score vs null.

# Weighted pooled f_align across all bins, weighting by N
num = (df["f_align"] * df["N"]).sum()
den = df["N"].sum()
f_pooled = num / den if den > 0 else np.nan

# We'll also estimate a null expectation by pooling the null means
null_num = (df["f_null_mean"] * df["N"]).sum()
null_den = den
f_null_pooled = null_num / null_den if null_den > 0 else np.nan

# Rough pooled sigma from null_std / sqrt(N_total)
sigma_eff = df["f_null_std"].mean() / np.sqrt(den)
z_score = (f_pooled - f_null_pooled) / (sigma_eff + 1e-12)

print(f"Total N ≈ {den:,}")
print(f"Pooled f_align ≈ {f_pooled:.3f}")
print(f"Pooled null mean ≈ {f_null_pooled:.3f}")
print(f"Approx z-score vs null ≈ {z_score:.2f} σ")
```

## Interpretation

* If this reproduces ~f_align ≈ 0.62–0.64 with N ~1.3M, and the z-score comes out very high (≫ 5σ is typical), then you’ve matched the public claim.
* Those numbers are what feed into the v2.0 summary: "pooled f ≈0.625 (N≈1.3M; p<1e-6)."
